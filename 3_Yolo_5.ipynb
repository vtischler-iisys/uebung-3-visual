## Load the model
`yolov5s` = the small version, downloaded via torch hub.

```python
import torch

model = torch.hub.load("ultralytics/yolov5", "yolov5s")
```

## Detect
Input can be a URL, file path, PIL/OpenCV image, numpy array, or a list of these.

```python
img = "https://ultralytics.com/images/zidane.jpg"
results = model(img)

results.print()
results.show()
```

Now our own image:

```python
img = "images/Hase.png"
results = model(img)

results.print()
results.show()
```
