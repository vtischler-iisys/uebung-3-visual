## Load the model

```python
from ultralytics import YOLOWorld
import matplotlib.pyplot as plt
import cv2

model = YOLOWorld("yolov8s-worldv2.pt")
```

## Helper to show the result

```python
def display_saved_image(file_path):
    img = cv2.imread(file_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    plt.imshow(img_rgb)
    plt.axis("off")
    plt.show()
```

## Pick classes and detect
Low `conf` = also show unsure boxes.

```python
model.set_classes(["tie", "hat", "flower"])

results = model.predict("images/Hase.png", max_det=100, iou=0.01, conf=0.01)
results[0].save(filename="result.jpg")
display_saved_image("result.jpg")
```

## Different classes, same image

```python
model.set_classes(["tie", "house", "person"])

results = model.predict("images/Hase.png", max_det=100, iou=0.01, conf=0.01)
results[0].save(filename="result.jpg")
display_saved_image("result.jpg")
```
